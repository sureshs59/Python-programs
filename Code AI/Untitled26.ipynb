```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target
df['species'] = iris.target_names[df['target']]

# Display the first few rows
print("Dataset head:")
display(df.head())

# --- Univariate Analysis ---
print("\n--- Univariate Analysis (Sepal Length) ---")
# Histogram and KDE plot of sepal length
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='sepal length (cm)', kde=True)
plt.title('Distribution of Sepal Length')
plt.show()

# Box plot of sepal length
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, y='sepal length (cm)')
plt.title('Box Plot of Sepal Length')
plt.show()

# --- Bivariate Analysis ---
print("\n--- Bivariate Analysis (Sepal Length vs. Sepal Width) ---")
# Scatter plot of sepal length vs. sepal width
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='sepal length (cm)', y='sepal width (cm)', hue='species')
plt.title('Sepal Length vs. Sepal Width by Species')
plt.show()

# Pairplot for selected features
print("\n--- Bivariate Analysis (Pairplot of Selected Features) ---")
sns.pairplot(df, vars=['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)'], hue='species')
plt.suptitle('Pairwise Relationships of Iris Features by Species', y=1.02)
plt.show()


# --- Multivariate Analysis ---
print("\n--- Multivariate Analysis (Heatmap of Correlation Matrix) ---")
# Correlation matrix
correlation_matrix = df[['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']].corr()

# Heatmap of the correlation matrix
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Iris Features')
plt.show()

# 3D Scatter plot (example - requires matplotlib in a different way for 3D)
# This is a more advanced example and might require different setup in some environments
# print("\n--- Multivariate Analysis (3D Scatter Plot - Sepal Length, Sepal Width, Petal Length) ---")
# fig = plt.figure(figsize=(10, 7))
# ax = fig.add_subplot(111, projection='3d')
#
# colors = ['blue', 'red', 'green']
# for i, species_name in enumerate(iris.target_names):
#     xs = df['sepal length (cm)'][df['species'] == species_name]
#     ys = df['sepal width (cm)'][df['species'] == species_name]
#     zs = df['petal length (cm)'][df['species'] == species_name]
#     ax.scatter(xs, ys, zs, c=colors[i], marker='o', label=species_name)
#
# ax.set_xlabel('Sepal Length (cm)')
# ax.set_ylabel('Sepal Width (cm)')
# ax.set_zlabel('Petal Length (cm)')
# plt.title('3D Scatter Plot of Iris Features by Species')
# plt.legend()
# plt.show()
```
